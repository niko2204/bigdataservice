## 학습목표

- 임베딩을 데이터에서 벡터로 가는 함수로 설명한다.
- 거리와 코사인 유사도를 손으로 계산한다.
- 동일한 벡터도 측정 방법에 따라 관계가 다르게 보일 수 있음을 설명한다.


```python
import math
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
pd.set_option("display.max_colwidth", 100)

def cosine(a, b):
    a, b = np.asarray(a, dtype=float), np.asarray(b, dtype=float)
    denom = np.linalg.norm(a) * np.linalg.norm(b)
    return float(a @ b / denom) if denom else 0.0
```

## 핵심 수학

유클리드 거리와 코사인 유사도는 서로 다른 질문에 답한다.

$$d(\mathbf{x},\mathbf{y})=\sqrt{\sum_i(x_i-y_i)^2}$$

$$\cos(\theta)=\frac{\mathbf{x}\cdot\mathbf{y}}{\|\mathbf{x}\|\|\mathbf{y}\|}$$


```python
def euclidean(a, b):
    a, b = np.asarray(a, dtype=float), np.asarray(b, dtype=float)
    return np.sqrt(np.sum((a - b) ** 2))

def cosine(a, b):
    a, b = np.asarray(a, dtype=float), np.asarray(b, dtype=float)
    denom = np.linalg.norm(a) * np.linalg.norm(b)
    return float(a @ b / denom) if denom else 0.0

vectors = {
    "해양 관광": np.array([0.9, 0.8]),
    "섬 여행": np.array([0.8, 0.9]),
    "배터리 공정": np.array([-0.7, 0.2]),
    "조선 제조": np.array([-0.8, 0.1]),
}
rows = []
for a in vectors:
    for b in vectors:
        if a < b:
            rows.append([a, b, euclidean(vectors[a], vectors[b]), cosine(vectors[a], vectors[b])])
pd.DataFrame(rows, columns=["항목A", "항목B", "거리", "코사인 유사도"]).round(3)
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
for label, v in vectors.items():
    ax.scatter(v[0], v[1], s=80)
    ax.annotate(label, v + 0.03)
ax.axhline(0, color="gray", lw=0.5); ax.axvline(0, color="gray", lw=0.5)
ax.set(title="장난감 임베딩 공간", xlabel="차원 1", ylabel="차원 2")
plt.show()
```
