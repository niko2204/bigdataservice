```python
import pandas as pd
url = 'https://raw.githubusercontent.com/OWNER/bigdataservice/main/data/sample/location_features.csv'
# GitHub 게시 전에는 아래 로컬 경로를 사용하세요.
df = pd.read_csv('../data/sample/location_features.csv')
df.head()
```

```python
df['카페당20대인구'] = (df['20대인구'] / (df['카페수'] + 1)).round(1)
df.sort_values('카페당20대인구', ascending=False)[['행정동명','카페당20대인구']].head()
```
