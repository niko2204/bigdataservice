# 03. 전처리와 EDA

```python
import pandas as pd
import plotly.express as px
df = pd.read_csv('../data/sample/location_features.csv')
display(df.isna().sum())
```

```python
summary = df.groupby('행정동명', as_index=False).agg(유동인구=('유동인구','sum'), 카페수=('카페수','sum'))
px.scatter(summary, x='카페수', y='유동인구', hover_name='행정동명', title='카페 수와 유동인구')
```
