## 학습목표
합, 평균, 편차, 제곱거리의 의미를 손으로 계산하고 NumPy 결과와 비교한다.

평균: x_bar = (1/n) sum(x_i), 편차: d_i = x_i - x_bar

유클리드 거리: d(x,y) = sqrt(sum((x_j-y_j)^2))

```python
import numpy as np
x = np.array([8, 10, 12, 14, 16], dtype=float)
print('합:', x.sum(), '개수:', x.size, '평균:', x.mean())
deviation = x - x.mean()
print('편차:', deviation, '편차의 합:', deviation.sum())
```
