# Lab 05. 회귀분석과 예측 오차
선형회귀는 예측값 y_hat = beta_0 + beta_1 x_1 + ... + beta_p x_p로 나타낸다. 최소제곱법은 SSE=sum((y-y_hat)^2)를 최소화한다.

MAE=mean(|y-y_hat|), RMSE=sqrt(mean((y-y_hat)^2)). RMSE는 큰 오차에 더 민감하다.
