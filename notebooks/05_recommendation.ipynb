# 05. 입지 추천 모델
방향을 통일한 후 정규화하고 가중합 점수를 계산합니다.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
df = pd.read_csv('../data/sample/location_features.csv')
raw = pd.DataFrame({'잠재고객':df['20대인구'],'유동인구':df['유동인구'],'경쟁완화':-df['카페수'],'접근성':df['주차장수'],'비용효율':-df['평균거래가']})
scaled = pd.DataFrame(MinMaxScaler().fit_transform(raw), columns=raw.columns)
weights = {'잠재고객':.25,'유동인구':.25,'경쟁완화':.20,'접근성':.15,'비용효율':.15}
df['적합도'] = sum(scaled[c]*w for c,w in weights.items())*100
df.sort_values('적합도', ascending=False)[['행정동명','적합도']].head()
```
