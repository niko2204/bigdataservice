## 2. 빅데이터 서비스란 무엇인가

데이터 분석은 “무슨 일이 있었는가?”에 답하는 과정이고, 데이터 서비스는 분석을 반복 가능한 기능으로 제공한다. 예를 들어 카페 수를 그래프로 한 번 그리는 것은 분석이다. 사용자가 업종과 중요도를 입력할 때마다 추천 지역과 근거가 갱신된다면 서비스다.

서비스는 다음 다섯 층으로 나눌 수 있다.

| 층 | 질문 | 이 프로젝트의 예 |
|---|---|---|
| 사용자 | 누가 어떤 결정을 하는가? | 예비 창업자가 후보 지역을 비교 |
| 데이터 | 어떤 관측값이 필요한가? | 인구, 점포, 주차장, 거래가 |
| 처리 | 어떻게 품질을 확보하는가? | 중복 제거, 행정동 표준화 |
| 분석·모델 | 어떻게 근거를 만드는가? | EDA, 정규화, 가중합 점수 |
| 인터페이스 | 어떻게 전달하는가? | Streamlit 표·차트·지도 |

## 3. 문제 정의 방법

좋은 문제 정의는 다음 문장을 완성한다.

> `[사용자]`가 `[상황]`에서 `[데이터]`를 근거로 `[결정]`하도록 돕는다.

예시: “목포에서 소규모 카페를 준비하는 예비 창업자가 인구·유동인구·경쟁·접근성 데이터를 근거로 비교할 행정동을 선택하도록 돕는다.”

“목포 상권을 분석한다”는 범위와 성공 기준이 없기 때문에 부족하다. 추천 정확도를 직접 측정하기 어렵다면 다음과 같은 대리 성공지표를 사용한다.

- 모든 추천에 수치 근거가 표시되는가?
- 가중치를 바꾸면 결과가 일관되게 갱신되는가?
- 처음 사용하는 사람이 3분 안에 후보지 3곳을 비교하는가?
- 새 데이터로 전체 파이프라인을 다시 실행할 수 있는가?

## 4. 요구사항 작성

기능 요구사항은 서비스가 해야 할 행동이다.

- 사용자는 업종을 선택할 수 있다.
- 사용자는 다섯 지표의 가중치를 바꿀 수 있다.
- 시스템은 추천 순위와 이유를 표시한다.

품질 요구사항은 행동이 얼마나 잘 수행되어야 하는지 정의한다.

- 같은 입력은 같은 결과를 생성해야 한다.
- API 키가 코드와 로그에 노출되지 않아야 한다.
- 데이터가 없을 때 오류 대신 안내 문구를 보여야 한다.
- 출처와 기준일을 화면에서 확인할 수 있어야 한다.

## 6. Git 협업 최소 규칙

- 작업 하나마다 작은 브랜치를 만든다: `feature/add-map-filter`.
- 커밋은 완료한 행동으로 쓴다: `Add business type filter`.
- Pull Request에는 변경 이유, 화면, 검증 방법을 기록한다.
- 다른 팀원이 검토한 뒤 병합한다.
- 데이터 원본, API 키, 가상환경은 커밋하지 않는다.

## 7. 자주 발생하는 문제
